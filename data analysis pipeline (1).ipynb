```python
def load_data(folder_path,file_name):
    #import pandas as pd
    if file_name.endswith('.csv'):
        df=pd.read_csv(folder_path+ "/" +file_name)
        print("You can access csv file by df keyword")
        return df
    elif file_name.endswith('.xlsx'):
        totalsheet=pd.ExcelFile(folder_path+ "\\" +file_name).sheet_names                   #sheet_names
        sheets={}
        for i in totalsheet:
            sheets.update([(i,pd.read_excel(folder_path+ "\\" +file_name,sheet_name=i))])
        print("You can access excel file by sheets keyword")
        return sheets
    else:
        print("Extension not found")
```

```python
def data_information(df):
    print("Information of all columns :",df.info())
    print("The number of dulicates in the dataset :",df.duplicated().sum())
    print("The percentage of null values in each column of the dataset :",df.isnull().sum()/df.shape[0])
    print("Values count of all categorical Column :")
    for i in df.select_dtypes(include=['object','bool']):
        if (len(df[i].unique())/df.shape[0])<0.3:
            print(i,":",df[i].value_counts())
    print("Description of numerical columns in the dataset :\n",df.describe())
data_information(df)    
```

```python
def data_cleaning(df):
    #for changing datatype
    cleaning_type=input('Enter which type of cleaning that you want to perform(dtype conversion/null handling/miscellaneous) :')
    if cleaning_type=='dtype conversion':
        choice=='Yes'
        while choice.lower()=='yes':
            column_name=input("Give Column name for changing the Data type :")                 #(object, float )
            if df[column_name].dtype=='object':
                convertdtype=input("Give datatype for conversion:")
                if convertdtype=='float':
                    try:
                        df[column_name]=df[column_name].astype('float')
                    except (ValueError):
                        pass
                elif convertdtype=='datetime':
                    df[column_name]=pd.to_datetime(df[column_name],format='mixed',errors='coerce')
            choice=input("Do you want to change the datatype of another column (yes/no):")
    elif cleaning_type=='null handling':

        for i in df.columns[df.isnull().any()].tolist():
            if df[i].dtype=='float':
                print("Mean:", df[i].mean())
                print("Median:", df[i].median())
                print("Mode:", df[i].mode()[0])
                print("Min:", df[i].min())
                print("Max:", df[i].max())
                print("Std:", df[i].std())

                method = input("Method (mean/median/mode/value/bfill/ffill/drop): ").lower()

                if method == 'mean':
                    df[i] = df[i].fillna(df[i].mean())

                elif method == 'median':
                    df[i] = df[i].fillna(df[i].median())

                elif method == 'mode':
                    df[i] = df[i].fillna(df[i].mode()[0])

                elif method == 'value':
                    value = float(input("Enter value: "))
                    df[i] = df[i].fillna(value)

                elif method in ['bfill', 'ffill']:
                    df[i] = df[i].fillna(method=method)

                elif method == 'drop':
                    df = df.dropna(subset=[i])

                else:
                    print("Invalid method")

            # Object --- Category
            elif df[i].dtype == 'object':

                print(df[i].value_counts())

                method = input("Method (mode/value/drop/bfill/ffill): ").lower()  #bfill ffill

                if method == 'mode':
                    df[i] = df[i].fillna(df[i].mode()[0])

                elif method == 'value':
                    value = input("Enter value: ")
                    df[i] = df[i].fillna(value)
                    
                elif method in ['bfill', 'ffill']:
                    df[i] = df[i].fillna(method=method)
                    
                elif method == 'drop':
                    df = df.dropna(subset=[i])

                else:
                    print(" Invalid method")

            # Datetime
            elif 'datetime' in str(df[i].dtype):
                method = input("Method (bfill/ffill)/value: ").lower()
                if method in ['bfill','ffill']:
                    
                    df[i] = df[i].fillna(method=method)

                elif method == 'value':
                    value = input("Enter value: ")
                    df[i] = df[i].fillna(value)
                else:
                    print("error, choose from method option")
```

```python
def feature_engineering(df):
    feature_type=input("Enter feature engineering type (concat/datetime/case when/arithmetic) : ")

    if feature_type=='concat':
        choice='yes'
        while choice.lower()=='yes':
            column1=input("Enter first column name : ")
            column2=input("Enter second column name : ")
            new_column_name=input("Enter new column name : ")
            separator=input("Enter separator (for no separator just press enter) : ")

            if column1 in df.columns and column2 in df.columns:
                try:
                    df[new_column_name]=df[column1].astype(str)+separator+df[column2].astype(str)
                    print(f"{new_column_name} created successfully")
                except Exception as e:
                    print("Error occured :",e)
            else:
                print("Column not found")

            choice=input("Do you want to create another concat column (yes/no) : ")

    elif feature_type=='datetime':
        choice='yes'
        while choice.lower()=='yes':
            column_name=input("Enter datetime column name : ")

            if column_name in df.columns:
                try:
                    if not pd.api.types.is_datetime64_any_dtype(df[column_name]):
                        df[column_name]=pd.to_datetime(df[column_name],format='mixed',errors='coerce')

                    datetime_feature=input("Enter datetime feature (year/month/day/day_name/hour/minute/is_weekend) : ")
                    new_column_name=input("Enter new column name : ")

                    if datetime_feature=='year':
                        df[new_column_name]=df[column_name].dt.year

                    elif datetime_feature=='month':
                        df[new_column_name]=df[column_name].dt.month

                    elif datetime_feature=='day':
                        df[new_column_name]=df[column_name].dt.day

                    elif datetime_feature=='day_name':
                        df[new_column_name]=df[column_name].dt.day_name()

                    elif datetime_feature=='hour':
                        df[new_column_name]=df[column_name].dt.hour

                    elif datetime_feature=='minute':
                        df[new_column_name]=df[column_name].dt.minute

                    elif datetime_feature=='is_weekend':
                        df[new_column_name]=df[column_name].dt.dayofweek.isin([5,6]).astype(int)

                    else:
                        print("Invalid datetime feature")
                        continue

                    print(f"{new_column_name} created successfully")
                except Exception as e:
                    print("Error occured :",e)
            else:
                print("Column not found")

            choice=input("Do you want to create another datetime feature (yes/no) : ")

    elif feature_type=='case when':
        choice='yes'
        while choice.lower()=='yes':
            column_name=input("Enter column name : ")
            condition=input("Enter condition (greater than/less than/equal to/between) : ")
            new_column_name=input("Enter new column name : ")

            if column_name in df.columns:
                try:
                    if condition=='greater than':
                        value=float(input("Enter threshold value : "))
                        true_value=input("Enter value if condition is True : ")
                        false_value=input("Enter value if condition is False : ")
                        df[new_column_name]=df[column_name].apply(lambda x: true_value if x>value else false_value)

                    elif condition=='less than':
                        value=float(input("Enter threshold value : "))
                        true_value=input("Enter value if condition is True : ")
                        false_value=input("Enter value if condition is False : ")
                        df[new_column_name]=df[column_name].apply(lambda x: true_value if x<value else false_value)

                    elif condition=='equal to':
                        value=input("Enter value to compare : ")
                        true_value=input("Enter value if condition is True : ")
                        false_value=input("Enter value if condition is False : ")
                        df[new_column_name]=df[column_name].astype(str).apply(lambda x: true_value if x==value else false_value)

                    elif condition=='between':
                        lower=float(input("Enter lower value : "))
                        upper=float(input("Enter upper value : "))
                        true_value=input("Enter value if condition is True : ")
                        false_value=input("Enter value if condition is False : ")
                        df[new_column_name]=df[column_name].apply(lambda x: true_value if lower<=x<=upper else false_value)

                    else:
                        print("Invalid condition")
                        continue

                    print(f"{new_column_name} created successfully")
                except Exception as e:
                    print("Error occured :",e)
            else:
                print("Column not found")

            choice=input("Do you want to create another case when column (yes/no) : ")

    elif feature_type=='arithmetic':
        choice='yes'
        while choice.lower()=='yes':
            column1=input("Enter first column name : ")
            operator=input("Enter operator (+,-,*,/) : ")
            column2=input("Enter second column name : ")
            new_column_name=input("Enter new column name : ")

            if column1 in df.columns and column2 in df.columns:
                try:
                    if operator=='+':
                        df[new_column_name]=df[column1]+df[column2]

                    elif operator=='-':
                        df[new_column_name]=df[column1]-df[column2]

                    elif operator=='*':
                        df[new_column_name]=df[column1]*df[column2]

                    elif operator=='/':
                        df[new_column_name]=df[column1]/df[column2]

                    else:
                        print("Invalid operator")
                        continue

                    print(f"{new_column_name} created successfully")
                except Exception as e:
                    print("Error occured :",e)
            else:
                print("Column not found")

            choice=input("Do you want to create another arithmetic column (yes/no) : ")

    else:
        print("Invalid feature engineering type")

    return df
```
